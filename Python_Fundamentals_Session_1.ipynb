**3.Variables**

```python
#Static VS Dynamic Typing

#Static typing in java
int a=5  #static typing means you need to declare the type of the variable first.

#Dynamic typing in python
a= 'apurva' #dynamic typing means there is no need to declare the type of the variable.
```

```python
#Static and Dynamic Binding

#static binding
int a =5
print(a)
#a = 'apurva' it will give error as the data type of the variable has been changed
int a= 6
print(a) #this will be printed
```

```python
#dynamic binding
a=8
print(a)

a= 'apurva'
print(a)
```

```python
#stylish declaration techniques
a=1
b=2
c=3
print(a,b,c)
```

```python
a,b,c=1,2,3
print(a,b,c)
```

```python
a=b=c=1
print(a,b,c)
```
