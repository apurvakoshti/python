**6.Type Conversion**

```python
#Implicit and Explicit Conversion

#Implicit
x=4
y=4.5
result = x + y
print(result)

a=4
b='4'
result= a + b
print(result) #This will give error
```

```python
#Explicit (Type Casting)
#str -> int
b= int('4')
print(b)
type(b)
```

```python
#Solution to above example
fnum= int(input('Enter the first number: '))
snum= int(input('Enter the second number: '))

result= fnum+snum #if i would have used here like int(fnum) and print the type of fnum below, it would have given me as str
print(result)
type(fnum) #here it gives type as int
```
