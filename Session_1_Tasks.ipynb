Q.2:-Write a program that will convert celsius value to fahrenheit.

```python
# formula for converting the celsius to fahrenheit is f= c*9/5+32

c= int(input("Enter the temperature in Celsius: "))
f= c*(9/5)+32
print("The temperature in fahrenheit is: ",f)
```

Q.3:-Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

Q.4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

```python
x1=int(input("Enter x1: "))
x2=int(input("Enter x2: "))
y1=int(input("Enter y1: "))
y2=int(input("Enter y2: "))

distance= ((x2-x1)**2+ (y2-y1)**2)**0.5

print(distance)
```

Q.6:-Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.
For example: Input: heads -> 4 legs -> 12
Output: dogs -> 2 chicken -> 2

```python
heads= int(input("enter the total number of heads: "))
legs= int(input("enter the total number of legs: "))

# d + c = heads
# 4d + 2c = legs


dogs = (legs - 2 * heads) / 2
chickens = heads - dogs

print('Number of dogs: ', dogs)
print('Number of chicken: ', chickens
      )
```

Q.8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

Q.9:-Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

```python
a= int(input("enter the num1: "))
b= int(input("enter the den1: "))

c= int(input("enter the num2: "))
d= int(input("enter the den2: "))

#sum= a/b+c/d = a*d+c*b/b*d

new_num1= a*d
new_num2= c*b

com_den= b*d

sum_new= new_num1 + new_num2

print("sum: ", sum_new, "/", com_den)

```

Q.10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.

```python
import math

h= int(input("enter the height of milk tank: "))
w= int(input("enter the width of the milk tank: "))
b= int(input("enter the breadth of the milk tank: "))

r= int(input("enter the radius of the glass: "))
hg= int(input("enter the height of the glass: "))

volume_of_glass= math.pi * (r**2)*hg
volume_of_tank= h * w * b

no_of_milk= volume_of_tank//volume_of_glass

print(int(no_of_milk))
```
